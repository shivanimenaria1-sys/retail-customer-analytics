## 5. Missing Values
We calculate the total count and percentage of missing values for each column to determine which columns require imputation or deletion.

```python
# Calculate total missing values and their percentage per column
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage (%)': missing_percentage
})

# Display sorted missing values overview
missing_df.sort_values(by='Missing Count', ascending=False)
```

## 6. Duplicate Records
Check for duplicate rows in the dataset.

```python
# Check for duplicate records
duplicate_count = df.duplicated().sum()
print(f"Total duplicate records: {duplicate_count}")
```

## 7. Unique Values
Let's show the unique value count for each column and highlight which columns are categorical.

```python
# Show unique counts for every column and highlight categorical columns
unique_counts = df.nunique()
dtypes = df.dtypes

unique_summary = pd.DataFrame({
    'Unique Count': unique_counts,
    'Data Type': dtypes,
    'Is Categorical': dtypes.apply(lambda x: 'Yes' if x == 'object' or x.name == 'category' else 'No')
})

unique_summary.sort_values(by='Unique Count', ascending=False)
```

## 8. Data Types Summary
Count the number of columns by numeric, categorical, and datetime categories.

```python
# Count the number of columns by broad category
numeric_cols = df.select_dtypes(include=['number']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
datetime_cols = df.select_dtypes(include=['datetime', 'datetimetz']).columns.tolist()

print(f"Numeric Columns ({len(numeric_cols)}): {numeric_cols}\n")
print(f"Categorical Columns ({len(categorical_cols)}): {categorical_cols}\n")
print(f"Datetime Columns ({len(datetime_cols)}): {datetime_cols}\n")

print(f"Summary Count:")
print(f"- Numeric columns: {len(numeric_cols)}")
print(f"- Categorical columns: {len(categorical_cols)}")
print(f"- Datetime columns: {len(datetime_cols)}")
```

## 9. Initial Observations
Based on the initial inspection of the dataset, here is a summary of findings:

### Dataset Overview
* **Number of rows**: 2,240

### Data Quality & Missing Values
* **Missing values**: The `Income` column has **24 missing values** (representing ~1.07% of the total dataset). All other columns are complete.
* **Constant Columns (Zero Variance)**: The columns `Z_CostContact` (constant value: 3) and `Z_Revenue` (constant value: 11) have only 1 unique value and do not carry any predictive power.

* **Outliers**:
  * `Year_Birth` contains values indicating customers born in the late 1800s or early 1900s, which could result in unrealistic ages (e.g., >120 years).
  * `Income` has some extremely high values that should be checked for outliers.
